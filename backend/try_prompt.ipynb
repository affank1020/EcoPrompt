# Try the prompt complexity model

Select the project's `.venv` Python kernel, then run the two code cells. Edit `prompt` and rerun the second cell as often as you like. This notebook uses the saved model; it does not download data or train anything.

Classes: **easy (1–3)**, **medium (4–7)**, **hard (8–10)**.

```python
from pathlib import Path
import joblib
from complexity import classify_prompt

model_path = Path('artifacts/complexity_model.joblib')
if not model_path.is_file():
    raise FileNotFoundError(f'Saved model not found: {model_path.resolve()}. Run main.ipynb first.')
model = joblib.load(model_path)
print('Saved complexity model loaded.')
```

```python
prompt = "Build a complex website with multiple pages, a responsive design, and interactive elements."
result = classify_prompt(prompt, model=model)
print(f"{result['label'].capitalize()} (complexity {result['score_range']}) — confidence: {result['confidence']:.1%}")
```
